```python
import mailbox

emails_per_file = 5000  # 1ファイルあたりのメール数

file_count = 1
current_count = 0
target_mbox = None
out_path = "/home/yutaka/src_p/260504_Google_mail/output/"

for i, message in enumerate(mbox):
    if i % emails_per_file == 0:
        if target_mbox:
            target_mbox.close()
        new_filename = out_path + f'split_{file_count}.mbox'
        target_mbox = mailbox.mbox(new_filename)
        file_count += 1
        print(f'{new_filename} を作成中...')

    target_mbox.add(message)

if target_mbox:
    target_mbox.close()

print("すべての分割が完了しました。")
```
