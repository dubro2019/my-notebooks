```python
from email.header import decode_header

def decode_email_subject(subject_raw):
    if not subject_raw:  return "(No Subject)"

    # エンコードされたヘッダーをデコード
    decoded_parts = decode_header(subject_raw)

    # デコードされた部分を結合して完全な件名を作成(+エンコーディングの処理)
    subject_pieces = []
    for content, encoding in decoded_parts:

        if isinstance(content, bytes):
            # エンコーディングが不明な場合は utf-8 や iso-2022-jp を試みる
            encoding = encoding if encoding else 'iso-2022-jp'
            try:
                subject_pieces.append(content.decode(encoding))
            except (UnicodeDecodeError, LookupError):
                subject_pieces.append(content.decode('utf-8', errors='replace'))
        else:
            # すでに文字列の場合はそのまま追加
            subject_pieces.append(content)
    
    subject = "".join(subject_pieces)
    return subject
```

```python
from email.utils import parsedate_to_datetime
from datetime import timezone, timedelta

# 日本時間の定義 (+9時間)
JST = timezone(timedelta(hours=9))

def format_to_jst(date_str):
    dt = parsedate_to_datetime(date_str) #datetimeオブジェクトに変換
    dt_jst = dt.astimezone(JST) # 日本時間(JST)に変換
    return dt_jst.strftime('%Y/%m/%d %H:%M')

def format_no_timezone(date_str):
    dt = parsedate_to_datetime(date_str) #datetimeオブジェクトに変換
    return dt.strftime('%Y/%m/%d %H:%M')

```
